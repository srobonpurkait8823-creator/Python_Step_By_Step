```python
#Write a program to accept the cost price of a car and display the road tax to be paid according to the following criteria.
#Cost > 1000000 : 15%
#Cost 500000 <= 1000000 : 10%
#Cost <= 500000 : 5%

cost = float(input("Pleaser enter the cost of your vehicle"))
if(cost > 1000000):
  print("You have to pay 15% as road tax")
elif(cost > 500000 and cost <= 1000000):
  print("You have to pay 10% as road tax")
elif(cost <= 500000):
  print("You have to pay 5% as road tax")
else:
  print("Invalid input")
```

```python
# #Practice Program 1
# #Write a Python program to take two inputs from user(num1 and mu,2) and perform Addition, Subtraction, Multiplication and Division.

num_one = float(input("Enter the first number: "))
num_two = float(input("Enter the second number: "))
operator = input("Enter the operator: ")

if (operator == "+"):
  print("Addition: ", num_one + num_two)
elif(operator == "-"):
  print("Subtraction: ", num_one - num_two)
elif(operator == "*"):
  print("Multiplication: ", num_one * num_two)
elif(operator == "/"):
  print("Division: ", num_one / num_two)
else:
    print("Invalid operator entry!")
  #End
```
